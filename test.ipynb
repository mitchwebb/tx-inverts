```python
import pandas as pd
from shapely.geometry import Point
import geopandas as gpd
from pandas import DataFrame
from geopandas.geodataframe import GeoDataFrame
from backend.config.map_constants import TEXAS_GEOJSON
import csv
from pathlib import Path
import re
from backend.config.data import DATA_OUT_PATH
import os

filepath = os.path.join(DATA_OUT_PATH, 'big_occurrence.txt')
chunk_size = 1000000

filtered_chunks = []

observation_count = 0

# Get Texas boundary
texas_shape_file = TEXAS_GEOJSON
texas_gdf = gpd.read_file(texas_shape_file).to_crs('EPSG:4326')
texas_boundary = texas_gdf.geometry.union_all()

for chunk in pd.read_csv(
	filepath,
	delimiter='\t',
	quoting=csv.QUOTE_NONE,
	on_bad_lines='warn',
	low_memory=False,
	chunksize=chunk_size
):

	# STEP 1: FILTER GEOMETRIES
	# Drop missing coordinates
	chunk = chunk.dropna(subset=['decimalLongitude', 'decimalLatitude'])

	# Convert lon/lat to Point geometries
	geometry = [
		Point(xy) for xy in zip(chunk['decimalLongitude'],
								chunk['decimalLatitude'])
	]
	# Convert observations to GDF
	df = gpd.GeoDataFrame(chunk, geometry=geometry, crs='EPSG:4326')

	# Filter observations to Texas boundary
	df = df[df.geometry.within(texas_boundary)]
	bad_location_count = len(chunk) - len(df)
	if bad_location_count:
		print(
			f'Removed {bad_location_count} records found outside of Texas')

	df = pd.DataFrame(df)
 
	# Add newly filtered chunk to filtered_chunks
	filtered_chunks.append(df)
 
    # Combine all filtered chunks
	combined_df = pd.concat(filtered_chunks, ignore_index=True)
 
# Create cleaned directory if it doesn't exist
cleaned_dir = os.path.join(DATA_OUT_PATH, 'cleaned')

os.makedirs(cleaned_dir, exist_ok=True)

combined_df.to_csv(os.path.join(
	cleaned_dir, 'big_texas_obs.csv'), sep="\t", index=False)
 
```

```python
from backend.data_util.gbif.process_observations import _parse_ambiguous_match, AMBIGUOUS_DMY_PATTERN_REGEX, AMBIGUOUS_MDY_PATTERN_REGEX

test_cases = [
    # Month-only numeric
    ("05/2020", {"year": 2020, "month": 5, "day": None}),
    ("5/2020", {"year": 2020, "month": 5, "day": None}),

    # Month-only textual
    ("Jan 2020", {"year": 2020, "month": 1, "day": None}),
    ("December 2020", {"year": 2020, "month": 12, "day": None}),

    # Full MDY numeric
    ("12/25/2020", {"year": 2020, "month": 12, "day": 25}),   # unambiguous
    ("01/13/2020", {"year": 2020, "month": 1, "day": 13}),    # unambiguous (day >= 13)
    ("01/02/2020", False),                                    # ambiguous (could be DMY)

    # Full DMY numeric
    ("25/12/2020", {"year": 2020, "month": 12, "day": 25}),
    ("13/01/2020", {"year": 2020, "month": 1, "day": 13}),
    ("02/01/2020", False),

    # With suffix
    ("1st Jan 2020", {"year": 2020, "month": 1, "day": 1}),
    ("3rd/05/2020", {"year": 2020, "month": 5, "day": 3}),

    # Extra cases
    ("July 4th, 2020", {"year": 2020, "month": 7, "day": 4}),
    ("2020", False),                # year alone is not useful
    ("random text", False),         # no match at all
]

for text, expected in test_cases:
    mdy = AMBIGUOUS_MDY_PATTERN_REGEX.search(text)
    dmy = AMBIGUOUS_DMY_PATTERN_REGEX.search(text)
    result = None
    if dmy:
        # print(dmy.group('day'))
        parsed_dmy = _parse_ambiguous_match(dmy)
        if parsed_dmy:  # only accept if unambiguous
            result = parsed_dmy

    elif mdy:
        # print(mdy.group('day'))
        parsed_mdy = _parse_ambiguous_match(mdy)
        if parsed_mdy:  # fallback to MDY only if DMY failed
            result = parsed_mdy

    print(text, "→", result, "| expected:", expected)
```

```python
def build_lineages(df: pd.DataFrame) -> pd.DataFrame:
    df = df.copy()
    
    # Map taxon_id → row index for fast lookup
    id_to_idx = {tid: i for i, tid in enumerate(df["taxon_id"].values)}
    
    # Build adjacency (parent → children)
    children_map = defaultdict(list)
    in_degree = defaultdict(int)
    for child, parent in zip(df["taxon_id"].values, df["parent_name_usage_id"].values):
        if pd.notna(parent) and parent in id_to_idx:
            children_map[parent].append(child)
            in_degree[child] += 1
    
    # Start queue with root taxa
    queue = deque([tid for tid in df["taxon_id"].values if in_degree[tid] == 0])
    
    # Initialize lineage arrays (NumPy for speed)
    lineage = np.full((len(df), len(RANK_COLS)), np.nan, dtype="float64")
    
    while queue:
        tid = queue.popleft()
        i = id_to_idx[tid]
        rank = df.at[i, "taxon_rank"]
        parent = df.at[i, "parent_name_usage_id"]
        
        # If has parent, copy parent's lineage row
        if pd.notna(parent) and parent in id_to_idx:
            lineage[i, :] = lineage[id_to_idx[parent], :]
        
        # Overwrite own rank slot
        if rank in RANK_ORDER:
            lineage[i, RANK_ORDER.index(rank)] = tid
        
        # Push children
        for child in children_map.get(tid, []):
            queue.append(child)
    
    # Attach lineage arrays back to DataFrame
    for j, col in enumerate(RANK_COLS):
        df[col] = lineage[:, j].astype("Int64")  # nullable int
    
    return df
```
